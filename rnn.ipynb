```python
import pandas as pd
import torch
```

```python
"""

CSV
 ↓
Clean text
 ↓
Convert ham/spam → 0/1
 ↓
Tokenize messages
 ↓
Build vocabulary
 ↓
Convert words → numbers
 ↓
Pad sequences
 ↓
Train/test split
 ↓
RNN
 ↓
Spam / Not Spam

"""
```

```python
data = '/content/spam.csv'
```

```python
df = pd.read_csv('/content/spam.csv', encoding='latin1')
```

```python
df = df[['v1', 'v2']]
```

```python
df['v2']
```

```python
for i in range(len(df)):
  if df.loc[i,'v1'] == 'spam':
    df.loc[i,'v1'] = 1
  else:
    df.loc[i,'v1'] =0
```

```python
X = df['v2']
y= df['v1']
```

```python
X[0]
```

```python


tokenized = []

for message in X:
    words = message.split()
    tokenized.append(words)
```

```python
len(tokenized)
```

```python
all_words = []

for sentence in tokenized:
    for word in sentence:
        all_words.append(word)
```

```python
unique_words = list(set(all_words))
vocabulary = {}
for i in range(len(unique_words)):
  vocabulary[unique_words[i]] = i
```
